## Was ist Künstliche Intelligenz?

**KI** = Maschinen, die Aufgaben erledigen, die normalerweise menschliche Intelligenz erfordern.

### Beispiele im Alltag:
| Anwendung | Beispiel |
|-----------|----------|
| Sprachassistenten | Siri, Alexa |
| Empfehlungen | Netflix, Spotify |
| Übersetzung | DeepL, Google Translate |
| Chatbots | ChatGPT, Claude |

## Arten von KI

### Schwache KI (Narrow AI)
- Kann **eine** Aufgabe gut
- Beispiel: Schach-Computer, Spamfilter

### Starke KI (General AI)
- Könnte **alles** wie ein Mensch
- **Existiert noch nicht!**

## Experiment: Regelbasierter Chatbot

Dieser Chatbot nutzt **feste Regeln** - keine echte KI!

````python
def einfacher_chatbot(nachricht: str) -> str:
    """Ein regelbasierter Chatbot - keine echte KI!"""
    nachricht = nachricht.lower()
    
    if "hallo" in nachricht or "hi" in nachricht:
        return "Hallo! Wie kann ich dir helfen?"
    elif "wie geht" in nachricht:
        return "Mir geht es gut! Ich bin ein Programm."
    elif "was ist ki" in nachricht:
        return "KI = Künstliche Intelligenz"
    else:
        return "Das verstehe ich nicht."

# Teste den Chatbot:
print(einfacher_chatbot("Hallo!"))
print(einfacher_chatbot("Was ist KI?"))
print(einfacher_chatbot("Wie wird das Wetter?"))
````

## Traditionell vs. Machine Learning

```
Traditionell:  Programmierer schreibt Regeln → Computer folgt
ML:            Computer bekommt Beispiele → Computer lernt Regeln
```
