## Die drei Arten von Machine Learning

### 1. Supervised Learning (Überwachtes Lernen)
- Lernt aus **gelabelten** Daten
- Beispiel: E-Mails mit Label "Spam" oder "Kein Spam"

### 2. Unsupervised Learning (Unüberwachtes Lernen)
- Findet **selbst** Muster in Daten
- Beispiel: Kunden in Gruppen einteilen

### 3. Reinforcement Learning (Verstärkendes Lernen)
- Lernt durch **Belohnung/Bestrafung**
- Beispiel: Spiele lernen

## Experiment: Supervised Learning simulieren

Wir simulieren, wie ein Modell aus Beispielen lernt:

```python
# Trainingsdaten: (Größe in cm, Gewicht in kg) → Tierart
trainingsdaten = [
    ((30, 5), "Katze"),
    ((25, 4), "Katze"),
    ((60, 25), "Hund"),
    ((70, 30), "Hund"),
    ((35, 6), "Katze"),
    ((55, 20), "Hund"),
]

print("Trainingsdaten:")
print("Größe(cm) | Gewicht(kg) | Tier")
print("-" * 35)
for (groesse, gewicht), tier in trainingsdaten:
    print(f"   {groesse}      |     {gewicht}       | {tier}")
```

```python
# Einfaches "Modell" - lernt Durchschnittswerte
katzen = [(g, w) for (g, w), t in trainingsdaten if t == "Katze"]
hunde = [(g, w) for (g, w), t in trainingsdaten if t == "Hund"]

katze_avg_groesse = sum(g for g, w in katzen) / len(katzen)
hund_avg_groesse = sum(g for g, w in hunde) / len(hunde)

print(f"Gelerntes Muster:")
print(f"  Katzen sind ca. {katze_avg_groesse:.0f} cm groß")
print(f"  Hunde sind ca. {hund_avg_groesse:.0f} cm groß")

def vorhersage(groesse, gewicht):
    """Sagt vorher, ob Katze oder Hund"""
    grenze = (katze_avg_groesse + hund_avg_groesse) / 2
    return "Katze" if groesse < grenze else "Hund"

# Teste das Modell
print(f"\nVorhersagen:")
print(f"  40cm, 8kg → {vorhersage(40, 8)}")
print(f"  65cm, 28kg → {vorhersage(65, 28)}")
```

## Wichtige Begriffe

| Begriff | Erklärung | Beispiel |
|---------|-----------|----------|
| **Feature** | Eingabe-Eigenschaft | Größe, Gewicht |
| **Label** | Die Antwort/Kategorie | Katze, Hund |
| **Training** | Modell lernt aus Daten | fit() |
| **Vorhersage** | Modell gibt Antwort | predict() |

## Echtes ML mit Scikit-learn

Jetzt das gleiche mit einer echten ML-Bibliothek:

```python
from sklearn.tree import DecisionTreeClassifier
import numpy as np

# Daten vorbereiten
X = np.array([[30, 5], [25, 4], [60, 25], [70, 30], [35, 6], [55, 20]])  # Features
y = np.array(["Katze", "Katze", "Hund", "Hund", "Katze", "Hund"])       # Labels

# Modell erstellen und trainieren
modell = DecisionTreeClassifier()
modell.fit(X, y)

print("Modell trainiert!")

# Vorhersagen
neue_tiere = [[40, 8], [65, 28], [28, 3]]
vorhersagen = modell.predict(neue_tiere)

print("\nVorhersagen:")
for tier, pred in zip(neue_tiere, vorhersagen):
    print(f"  {tier[0]}cm, {tier[1]}kg → {pred}")
```
