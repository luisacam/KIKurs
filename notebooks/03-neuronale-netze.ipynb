## Wie funktioniert ein Neuron?

```
Eingaben × Gewichte → Summe → Aktivierung → Ausgabe
```

1. Jede Eingabe wird mit einem Gewicht multipliziert
2. Alle gewichteten Eingaben werden addiert
3. Eine Aktivierungsfunktion entscheidet die Ausgabe

````python
def relu(x):
    """ReLU: Gibt x zurück wenn positiv, sonst 0"""
    return max(0, x)

def neuron(eingaben, gewichte, bias):
    """Simuliert ein einzelnes Neuron"""
    # 1. Gewichtete Summe berechnen
    summe = sum(x * w for x, w in zip(eingaben, gewichte))
    
    # 2. Bias addieren
    summe += bias
    
    # 3. Aktivierung anwenden
    return relu(summe)

# Beispiel
eingaben = [1.0, 2.0, 3.0]
gewichte = [0.5, -0.3, 0.8]
bias = 0.1

ergebnis = neuron(eingaben, gewichte, bias)
print(f"Eingaben:  {eingaben}")
print(f"Gewichte:  {gewichte}")
print(f"Bias:      {bias}")
print(f"Ausgabe:   {ergebnis}")
````

## Schritt-für-Schritt Berechnung

Lass uns die Berechnung visualisieren:

````python
eingaben = [1.0, 2.0, 3.0]
gewichte = [0.5, -0.3, 0.8]
bias = 0.1

print("Schritt 1: Gewichtete Multiplikation")
for i, (x, w) in enumerate(zip(eingaben, gewichte)):
    print(f"  x{i+1} × w{i+1} = {x} × {w} = {x*w}")

summe = sum(x * w for x, w in zip(eingaben, gewichte))
print(f"\nSchritt 2: Summe = {summe}")

mit_bias = summe + bias
print(f"\nSchritt 3: + Bias = {summe} + {bias} = {mit_bias}")

ausgabe = relu(mit_bias)
print(f"\nSchritt 4: ReLU({mit_bias}) = {ausgabe}")
````

## Aktivierungsfunktionen

Verschiedene Funktionen für verschiedene Zwecke:

````python
import math

def relu(x):
    return max(0, x)

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

# Vergleich
werte = [-2, -1, 0, 1, 2]

print("  x   │  ReLU  │ Sigmoid")
print("──────┼────────┼─────────")
for x in werte:
    print(f" {x:3}  │  {relu(x):4}  │  {sigmoid(x):.3f}")
````

## Mini Neuronales Netz

Ein Netz mit 2 Eingaben, 2 versteckten Neuronen und 1 Ausgabe:

````python
import random
random.seed(42)

class MiniNetz:
    def __init__(self):
        # Zufällige Gewichte (normalerweise werden diese trainiert!)
        self.w1 = [[random.uniform(-1,1), random.uniform(-1,1)],  # Neuron 1
                   [random.uniform(-1,1), random.uniform(-1,1)]]  # Neuron 2
        self.b1 = [random.uniform(-1,1), random.uniform(-1,1)]
        
        self.w2 = [random.uniform(-1,1), random.uniform(-1,1)]   # Ausgabe
        self.b2 = random.uniform(-1,1)
    
    def forward(self, x1, x2):
        # Versteckte Schicht
        h1 = relu(x1*self.w1[0][0] + x2*self.w1[0][1] + self.b1[0])
        h2 = relu(x1*self.w1[1][0] + x2*self.w1[1][1] + self.b1[1])
        
        # Ausgabeschicht
        out = sigmoid(h1*self.w2[0] + h2*self.w2[1] + self.b2)
        return out

netz = MiniNetz()

print("Eingabe  →  Ausgabe")
print("-" * 20)
for x1, x2 in [(0,0), (0,1), (1,0), (1,1)]:
    out = netz.forward(x1, x2)
    print(f"({x1}, {x2})   →  {out:.3f}")
````
